```python
import graphviz

# Cria o diagrama de blocos
dot = graphviz.Digraph('AreaQuadrado', format='png')

# 1. Definição dos Nós (Blocos)
dot.node('A', 'Início', shape='box')
dot.node('B', 'Ler lado (em cm)', shape='trapezium')
dot.node('C', 'area = lado * lado', shape='box')
dot.node('D', 'Exibir area (em cm²)', shape='parallelogram')
dot.node('E', 'Fim', shape='box')

# 2. Conexões (Fluxo)
dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')
dot.edge('D', 'E')

# Exibe o diagrama
dot
```
